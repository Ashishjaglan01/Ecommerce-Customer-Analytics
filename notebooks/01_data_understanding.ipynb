```python
# Count orders by status
print(orders["order_status"].value_counts(dropna=False))
```

```python
# Investigate missing timestamps by order status
timestamp_cols = [
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date"
]

for col in timestamp_cols:
    orders[col] = pd.to_datetime(orders[col], errors="coerce")

missing_by_status = (
    orders.groupby("order_status")
    .agg(
        total_orders=("order_id", "count"),
        missing_approved=("order_approved_at",
                          lambda x: x.isna().sum()),
        missing_carrier=("order_delivered_carrier_date",
                         lambda x: x.isna().sum()),
        missing_delivered=("order_delivered_customer_date",
                           lambda x: x.isna().sum())
    )
)

print(missing_by_status)
```

```python
customer_mapping = (
    customers.groupby("customer_unique_id")["customer_id"]
    .nunique()
    .sort_values(ascending=False)
)

print("Unique customer identities:",
      customers["customer_unique_id"].nunique())

print("Identities linked to multiple customer IDs:",
      (customer_mapping > 1).sum())

print("\nExamples:")
print(customer_mapping[customer_mapping > 1].head(10))
```

```python

# Check whether order IDs are unique
print("Orders:")
print("Total rows:", len(orders))
print("Unique order IDs:", orders["order_id"].nunique())

# Check whether customer IDs in orders exist in customers
customer_ids_in_orders = orders["customer_id"].isin(
    customers["customer_id"]
)

print("\nCustomer relationship:")
print("Matching customer IDs:", customer_ids_in_orders.sum())
print("Unmatched customer IDs:", (~customer_ids_in_orders).sum())

# Check whether order IDs in order_items exist in orders
item_order_match = order_items["order_id"].isin(
    orders["order_id"]
)

print("\nOrder items relationship:")
print("Matching order-item rows:", item_order_match.sum())
print("Unmatched order-item rows:", (~item_order_match).sum())

# Check whether product IDs in order_items exist in products
product_match = order_items["product_id"].isin(
    products["product_id"]
)

print("\nProduct relationship:")
print("Matching order-item rows:", product_match.sum())
print("Unmatched product IDs:", (~product_match).sum())
```

```python

# 1. How many orders contain multiple items?
items_per_order = order_items.groupby("order_id").size()

print("Orders with multiple items:",
      (items_per_order > 1).sum())

print("Maximum items in one order:",
      items_per_order.max())

# 2. How many orders have multiple payment records?
payments_per_order = payments.groupby("order_id").size()

print("\nOrders with multiple payment records:",
      (payments_per_order > 1).sum())

print("Maximum payment records per order:",
      payments_per_order.max())

# 3. How many orders have multiple reviews?
reviews_per_order = reviews.groupby("order_id").size()

print("\nOrders with multiple review records:",
      (reviews_per_order > 1).sum())

print("Maximum review records per order:",
      reviews_per_order.max())

# 4. Validate seller relationships
seller_match = order_items["seller_id"].isin(sellers["seller_id"])

print("\nUnmatched seller IDs:",
      (~seller_match).sum())
```

## Phase 1: Data Understanding — Summary

### Dataset Overview
The Olist Brazilian E-Commerce dataset contains nine related CSV tables covering customers, orders, order items, payments, reviews, products, sellers, geolocation, and product category translations.

### Key Findings
- The orders table contains 99,441 orders with unique order IDs.
- The customers table contains 99,441 customer records and 96,096 unique customer identities.
- The order items table contains 112,650 rows.
- The dataset covers orders from September 2016 to October 2018.
- All tested customer, order, product, and seller relationships matched successfully.
- Some orders contain multiple items, payment records, or reviews.

### Data Quality Observations
- Missing order timestamps are associated with different order statuses and require careful interpretation.
- Some product attributes are missing.
- Review titles and comment messages contain many missing values.
- The geolocation table contains exact duplicate rows that require further investigation.
- Date columns were identified for conversion to datetime types.
- Directly joining item and payment records may inflate revenue totals.

### Next Steps
